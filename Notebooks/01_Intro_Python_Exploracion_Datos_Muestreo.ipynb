## 1.4 Listas

Una **lista** guarda varios valores en un solo objeto, en un orden específico. Se crean con corchetes `[]`.

````python
# Temperaturas registradas durante una semana (°C)
temperaturas_semana = [16.2, 17.8, 15.9, 18.4, 19.1, 17.0, 16.5]

print(temperaturas_semana)

print("Número de datos:", len(temperaturas_semana))
print("Primer valor:", temperaturas_semana[0])
print("Último valor:", temperaturas_semana[-1])
print("Del segundo al cuarto:", temperaturas_semana[1:4])
````

## 2.1 ¿Qué es un DataFrame?

En `pandas`, un **DataFrame** es una tabla de datos con filas y columnas, igual que una hoja de cálculo. Cada columna puede tener un tipo de dato distinto (texto, números, fechas, etc.) y cada fila normalmente representa una observación (por ejemplo, una muestra de roca o una estación).

## 2.2 Nuestro conjunto de datos: muestras de roca de campo

Imagina que un equipo de geología hizo una campaña de campo en distintos sitios de México, recolectando muestras de roca y midiendo su densidad, porosidad y dureza. Vamos a generar ese conjunto de datos (como si fuera el archivo que el equipo entregó) y guardarlo como un archivo `.csv`, el formato más común para compartir datos tabulares.

> En un caso real, tú recibirías directamente el archivo `.csv` y solo harías el paso de **lectura** de la sección 2.3. Aquí lo generamos nosotros primero, únicamente para tener datos con los que practicar.

````python
np.random.seed(42)  # para que todas/os obtengamos los mismos datos "aleatorios"

sitios = ["Sierra de Guadalupe", "Valle de Bravo", "Pico de Orizaba",
          "Cañón del Sumidero", "Desierto de los Leones", "Malinalco"]

n_muestras = 120

tipo_roca = np.random.choice(
    ["Ígnea", "Sedimentaria", "Metamórfica"],
    size=n_muestras,
    p=[0.4, 0.35, 0.25]
)

# La densidad y la porosidad dependen del tipo de roca (con variación aleatoria)
densidad = np.where(
    tipo_roca == "Ígnea", np.random.normal(2.9, 0.15, n_muestras),
    np.where(tipo_roca == "Sedimentaria", np.random.normal(2.4, 0.2, n_muestras),
             np.random.normal(2.75, 0.18, n_muestras))
)

porosidad = np.where(
    tipo_roca == "Sedimentaria", np.random.normal(18, 5, n_muestras),
    np.random.normal(6, 2.5, n_muestras)
)
porosidad = np.clip(porosidad, 0.5, None)  # la porosidad no puede ser negativa

dureza_mohs = np.round(np.random.uniform(2, 7, n_muestras), 1)
profundidad_m = np.round(np.random.uniform(0.5, 15, n_muestras), 1)
sitio_muestreo = np.random.choice(sitios, size=n_muestras)

df_rocas = pd.DataFrame({
    "id_muestra": [f"R-{i:03d}" for i in range(1, n_muestras + 1)],
    "sitio": sitio_muestreo,
    "tipo_roca": tipo_roca,
    "densidad_g_cm3": np.round(densidad, 2),
    "porosidad_pct": np.round(porosidad, 1),
    "dureza_mohs": dureza_mohs,
    "profundidad_m": profundidad_m
})

df_rocas.to_csv("muestras_rocas_campo.csv", index=False)
print("Archivo 'muestras_rocas_campo.csv' creado con", len(df_rocas), "muestras.")
````
